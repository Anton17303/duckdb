```python
con = duckdb.connect(str(RAIZ / "data/processed/indicadores.duckdb"), read_only=True)
tablas = [r[0] for r in con.execute("SELECT table_name FROM duckdb_tables() WHERE table_name LIKE 'ind_%' ORDER BY 1").fetchall()]
tablas
```

```python
ind = lambda t: con.execute(f"SELECT * FROM {t}").df()
v = ind("ind_01_viajes_mensuales")
pv = v.pivot(index="periodo", columns="taxi_type", values="viajes")
pv.plot(marker="o", figsize=(9, 4), title="Indicador 1 — viajes por mes", color={"yellow": "#f2b705", "green": "#2e8b57"}); plt.show()
pv
```
